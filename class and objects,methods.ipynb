```python
# M. Classes, Methods & Objects
# Usecase1: Airline Ticket Management System (Methods Only)
# Design a Python class called AirlineTicket that demonstrates ticket operations using methods only, without storing any attributes inside the class.
# The class should allow the following operations using methods:
# Display Ticket Details :  A method should accept passenger information such as name, age, source, destination, and print the details in a formatted style of "Passenger:name of age:age is travelling from:source to:destination”.
#
#
# Confirm Ticket :  A method should return the string "Confirmed" when called.
#
#
# Cancel Ticket :  A method should return the string "Cancelled" when called.
#
#
# Update Destination :  A method should accept the old destination and a new destination, print a message showing the updated destination, and return the new destination value.

class AirlineTicket():
    def display_ticket_details(self,name,age,source,destination):
        print(f"passenger:{name} of age:{age} is travelling from source:{source} to destination:{destination} ")

    def confirm_ticket(self):
        return "confirmed"
    def cancel_ticket(self):
        return "cancelled"

    def update_destination(self,old_destination,new_destination):
        print(f"updated destination from {old_destination} to new destination {new_destination}")

t=AirlineTicket()

t.display_ticket_details("indhu",23,"chennai","bangalore")
print(t.confirm_ticket())
print(t.cancel_ticket())
t.update_destination("bangalore","america")
```
